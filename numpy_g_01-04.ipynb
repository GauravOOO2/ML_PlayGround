```python
import numpy as np

array = np.array([1,2,3,4,5])
print(array)
print(type(array))
```

```python
# this is a 2D array

array_2d = np.array([[1,2,3,4],[5,6,7,8]])

# this is 3D array 

array_3d = np.array([[[1,2,3],[5,6,7]],[[9,0,0],[5,2,0]]])

# this gives u the dimensions of the array 

print(array_2d.ndim)
print(array_3d.ndim)

higher_dim = np.array([1,2,3,4], ndmin=5)

print("high dim array",higher_dim)
```

```python
# time to slice 

array = np.array([1,2,3,4])

print(array[0] + array[2])

# slicing above 2d array

print(array_2d[1,1])
print(array_3d[0,1,2])
```
